# Quantized MLP to SystemVerilog

Course scaffold with completed parameter extraction functions in `utils/utils.py`.
Run from this `generator/` directory. Generated files are written to `../rtl/`.
The original Intel Neural Compressor loading path needs the course-compatible environment; see `../docs/reproduction.md`.
Stored 2024 execution output has been cleared for this portfolio copy.


```python
from model_trainer.trainer import LitTrainer
from neural_compressor.utils.pytorch import load
from utils.utils import QuantModelStatistic


def sv_generator(layers, model_name, model_path):
    model = LitTrainer(model_name)
    model = load(model_path, model)
    statistic = QuantModelStatistic(model, model_name)
    length = len(layers)
    layers = [None] + layers
    for i in range(length):
        rendered = statistic.render(last_layer=layers[i],
                                    current_layer=layers[i + 1],
                                    bit_width=8,
                                    relu=True,
                                    module=layers[i + 1].upper())
        with open('../rtl/{}.sv'.format(layers[i + 1].upper()), 'w') as file:
            file.write(rendered)


layers = ['fc1', 'fc2']
sv_generator(layers, 'MLP', './output')
```
